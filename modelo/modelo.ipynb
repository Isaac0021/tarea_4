```python
import pandas as pd
from sklearn.ensemble import RandomForestClassifier
from sklearn.model_selection import train_test_split
from sklearn.metrics import classification_report, roc_auc_score
import joblib
```

```python
df = pd.read_csv("/Users/isaaccastillo/Documents/Tarea_4/data/cardio_train.csv", sep=";")
df.head()
```

# Limpieza

```python
df["age"] = (df["age"] / 365).astype(int)
df["gender"] = df["gender"].map({1: 0, 2: 1})
df["cholesterol"] = df["cholesterol"].apply(lambda x: 1 if x > 1 else 0)
df["gluc"] = df["gluc"].apply(lambda x: 1 if x > 1 else 0)
df.head()
```

# Modelo y entrenamiento

```python
X = df.drop(["id", "cardio"], axis=1)
y = df["cardio"]

X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.4, random_state=42
)

# 2. Define a tuned RandomForestClassifier
model = RandomForestClassifier(
    n_estimators=300,         # increased number of trees
    max_depth=10,             # prevent overfitting
    min_samples_split=10,     # require more samples to split
    class_weight="balanced",  # handle potential class imbalance
    random_state=42
)

# 3. Train the model
model.fit(X_train, y_train)

# 4. Predict and evaluate
y_pred = model.predict(X_test)
y_proba = model.predict_proba(X_test)[:, 1]

# 5. Print classification report and AUC score
print(classification_report(y_test, y_pred))
print("ROC AUC Score:", roc_auc_score(y_test, y_proba))
```

# Guardar modelo entrenado

```python
joblib.dump(model, "../app/modelo.pkl")
```
